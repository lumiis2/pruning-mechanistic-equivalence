# Experimento 4 — One-shot magnitude pruning

**Hipótese.** A 50% de sparsity, o momento de seleção da máscara pode alterar a capacidade de escapar do shortcut após rewinding para a mesma inicialização.

**Resultado.** Máscaras shortcut, transition e robust escaparam em 3/3 seeds e terminaram robustas. Máscaras aleatórias não escaparam em nenhuma seed e tiveram `gap` final médio ≈0,587. Existe sinal de que a estrutura selecionada pelo treinamento importa, mesmo em sparsity moderada.
