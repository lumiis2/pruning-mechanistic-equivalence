# Experimento 5 — Sweep one-shot de sparsity

**Pergunta.** Em qual sparsity aparecem diferenças entre máscaras random, shortcut, transition e robust?

**Resultado.** A separação começa claramente em 80%: random não escapa, shortcut falha em 1/3 e transition/robust escapam em 3/3. Em 90%, random não escapa, shortcut escapa em 1/3 e transition/robust em 2/3. O regime crítico está entre 80% e 90%, com forte variação entre seeds.

```python
from pathlib import Path
import os,subprocess,sys,pandas as pd
from IPython.display import Image,display
ROOT=Path.cwd().resolve(); ROOT=ROOT.parent if ROOT.name=='notebooks' else ROOT
if not (ROOT/'src').exists() and Path('/content/src').exists(): ROOT=Path('/content')
os.chdir(ROOT); sys.path.insert(0,str(ROOT))
RUN_EXPERIMENT=False
if RUN_EXPERIMENT:
    subprocess.run([sys.executable,'scripts/run_sparsity_sweep.py','--config','configs/sanity_check.yaml'],check=True)
```

```python
summary=pd.read_csv('outputs/sanity_check/sparsity_sweep/sweep_summary.csv')
summary['escaped']=summary.escape_epoch.notna()
display(summary.groupby(['sparsity','condition']).agg(seeds=('seed','count'),escaped=('escaped','sum'),final_gap=('final_shortcut_gap','mean'),conflicting=('final_conflicting_accuracy','mean')).reset_index())
display(Image(filename='outputs/sanity_check/sparsity_sweep/sparsity_sweep.png'))
```
