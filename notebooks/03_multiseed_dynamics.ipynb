# Experimento 3 — Dinâmica densa em múltiplas seeds

**Hipótese.** A rede aprende cor primeiro e depois pode transicionar para forma, com momento dependente da seed. Checkpoints funcionais substituem comparações por época fixa.

**Resultado.** Todas as 5 seeds adquiriram o shortcut na época 1. Seeds 42, 45 e 46 ficaram robustas; seed 44 terminou parcialmente robusta (`conflicting=0.8965`) e seed 43 permaneceu fortemente dependente da cor (`conflicting=0.305`, `gap=0.695`). A variabilidade entre seeds é parte central do fenômeno.

```python
summary=pd.read_csv('outputs/sanity_check/multi_seed/seed_summary.csv')
display(summary)
history=pd.read_csv('outputs/sanity_check/multi_seed/all_seeds_history.csv')
for seed,rows in history.groupby('seed'):
    plt.plot(rows.epoch,rows.test_balanced_shortcut_gap,label=f'seed {seed}')
plt.axhline(.2,color='black',ls='--'); plt.ylim(-.05,1.05); plt.xlabel('Epoch'); plt.ylabel('Shortcut gap'); plt.legend(); plt.grid(alpha=.2); plt.show()
```
